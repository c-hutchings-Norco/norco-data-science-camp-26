A distribution is a group of all of the possible values that can occur for a given variable and the frequency at which the values occur. Like variables, distributions can be quantitative or categorical. We use different types of graphs to represent different types of distributions.


## B. Quantitative Distributions

1. Where is the center/middle of the distribution?
2. How spread out are the values in the distribution?
3. What shape does the distribution make? (Does it look the same on either side, or does one side look different?)

Scatterplots, histograms, and boxplots help us answer these questions.

## D. Making Comparisons

So far, we've looked at one variable at a time. This has allowed us to build a sense of familiarity with our data, but there are more interesting questions arise when we compare two groups together. For example:

- Do people who exercise sleep more than people who don't? 
- Does the number of piercings look different across genders? 
- Are people who think that pineapple belongs on pizza more likely to be born in some months than others?

To answer questions like these, we look at the distribution of one variable separately for each category of another.

We'll show the graphs side by side (in different colors) to make the comparisons easier. Instead of flipping between separate graphs, we can quickly see whether one group tends to have higher values, more spread, or a different shape.

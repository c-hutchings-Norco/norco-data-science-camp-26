### D.1. Task 6: Understanding Torr and PSI

From [Wikipedia](https://en.wikipedia.org/wiki/Atmosphere_of_Earth), we see the Earth's atmosphere consists of 3 primary gases and is around 78% nitrogen, 21% oxygen, 1% argon. There are also other gasses, most notably our frenemy carbon dioxide, which represents roughly 0.04% of our atmosphere.



The total pressure at sea level is around 14.7 pounds per square inch (PSI), meaning that the atmosphere is pushing every square inch of your body with that much force. We're used to it, so we don't notice.

There are many different units for pressure that are used instead of PSI. For example, the ISS Mimic website uses torr. One PSI is the same as roughly 51.7 torr.

### D.2. Task 7: Understanding Partial Pressure

Above, we saw the total pressure of the Earth's atmosphere is around 760 torr.

There is also a concept known as the "partial pressure", i.e. the pressure due to just one gas.

For example, the partial pressure of oxygen in Earth's atmosphere is 21% * 760 = 160 torr.

To get the partial pressure for a particular gas, we simply take the total pressure and multiply it by the appropriate fraction.

### D.3. Reflections on Partial Pressures for Nitrogen and Oxygen

Below we summarize what we've learned so far:

```python
place = ["Earth", "ISS", "Earth", "ISS"]
gas = ["Nitrogen", "Nitrogen", "Oxygen", "Oxygen"]
torr = [590, 570, 160, 160]

df = pd.DataFrame({
    'place': place,
    'gas': gas,
    'torr': torr
})
px.bar(df, x = "gas", color = "place", y = "torr", barmode = "group")
```

We see that the partial pressures of nitrogen and oxygen on the ISS are very similar to Earth.

Interestingly, our bodies don't need us to breathe nitrogen at all. Any partial pressure of nitrogen between 0 and hundreds of torr would be fine. Note that if you breathe air where the partial pressure of nitrogen exceeds 2000 torr [nitrogen narcosis](https://en.wikipedia.org/wiki/Nitrogen_narcosis).


Similarly, the partial pressure of oxygen is quite similar to Earth's. This is important because humans don't function when the partial pressure of oxygen is a bit below around a third of what we're used to. For example, at the peak of Everest, the partial pressure of Oxygen is around 50 torr. This is simply not enough oxygen, meaning that this altitude is known as ["the Death zone"](https://en.wikipedia.org/wiki/Death_zone).
